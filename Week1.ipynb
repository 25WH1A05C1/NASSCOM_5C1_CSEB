FUNCTIONS

```python
#1
def greet():
    print("Hello everyone")
greet()
```

```python
#2
def greet_user(name):
    print(f"Hello, {name}!")

greet_user("anju")
greet_user("chinnu")
```

```python
#3
def greet_user(name="Guest"):
    print(f"Welcome, {name}!")

greet_user()
greet_user("Anjali")
```

```python
#4
def print_names(*names):
    print("Names passed:", names)
    for name in names:
        print("Hello", name)

print_names("Anju", "bunny", "Chinni")
```

```python
#5
def print_info(**info):
    print("User Info:")
    for key, value in info.items():
        print(f"{key} = {value}")

print_info(name="Arun", age=25, city="Delhi")
```

```python
#6
def show_details(*args, **kwargs):
    print("Args:", args)
    print("Kwargs:", kwargs)

show_details(1, 2, 3, name="Raju", city="hyd")
```

```python
#7
def multiply(a, b):
    return a * b

def square(n):
    return multiply(n, n)

print("Square of 4:", square(4))
```

```python
numbers = [1, 2, 3, 4, 5, 6]
evens = list(filter(lambda x: x % 2 == 0, numbers))

print(evens)

numbers = [1, 2, 3, 4]
squared = list(map(lambda x: x ** 2, numbers))

print(squared)
```

```python
numbers = [1, 2, 3, 4, 5, 6]


squares = map(lambda x: x**2, numbers)

squared_evens = list(filter(lambda x: x % 2 == 0, squares))

print(squared_evens)

```

DICTIONARIES


```python
student = {
    'name': 'Alice',
    'age': 21,
    'course': 'Python'
}
print("Dictionary:", student)
```

```python
print("Student name:", student['name'])
```

```python
student['grade'] = 'A'
print("After adding grade:", student)


del student['age']
print("After deleting age:", student)

for key, value in student.items():
    print(f"{key}: {value}")
```

 LIST INDEXING

```python
my_list = [10, 20, 30, 40, 50]
```

```python
print("First element:", my_list[0])   # 10
print("Last element:", my_list[-1])   # 50
```

```python
print("Elements from index 1 to 3 (not inclusive):", my_list[1:4])  # [20, 30, 40]
print("From start to index 2:", my_list[:3])  # [10, 20, 30]
print("From index 2 to end:", my_list[2:])    # [30, 40, 50]
print("Every second element:", my_list[::2])  # [10, 30, 50]
print("Reversed list:", my_list[::-1])
```

TUPLE INDEXING

```python
my_tuple = ("a", "b", "c", "d", "e")
```

```python
print("Element at index 0:", my_tuple[0])
print("Element at index -2:", my_tuple[-2])
```

```python
print("Slice from index 1 to 3:", my_tuple[1:4])
print("Tuple reversed:", my_tuple[::-1])
```

STRING INDEXING

```python
text = "hello world"
```

```python
print("Character at index 0:", text[0])
print("Last character:", text[-1])
```

```python
print("First 6 characters:", text[:6])   # 'Python'
print("From 6 to end:", text[6:])        # 'Programming'
print("Every 2nd character:", text[::2])
print("Reversed string:", text[::-1])
```
