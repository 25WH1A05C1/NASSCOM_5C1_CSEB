INTEGRATED CASE STUDY ASSIGNMENT

 Part A – Python Programming

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


data = {
    "Student": ["S1", "S2", "S3", "S4", "S5",
                "S6", "S7", "S8", "S9", "S10"],

    "Study_Hours": [2, 3, 4, 5, 6, 7, 8, 4, 6, 9],

    "Attendance": [65, 70, 75, 80, 85, 90, 92, 78, 88, 95],

    "Assignment_Score": [55, 60, 65, 72, 78, 85, 88, 70, 80, 92],

    "Previous_Score": [58, 62, 68, 70, 76, 82, 86, 72, 79, 90],

    "Final_Score": [60, 64, 70, 75, 82, 88, 91, 74, 84, 94]
}

df = pd.DataFrame(data)
print(df)
```

```python
print("Number of students:", len(df))

print("\nColumn names:")
print(df.columns)

print("\nData types:")
print(df.dtypes)

print("\nFirst five records:")
print(df.head())

print("\nLast five records:")
print(df.tail())
```

Part B – Pandas Data Analysis

```python
print("Average Study Hours:",
      df["Study_Hours"].mean())

print("Average Attendance:",
      df["Attendance"].mean())

print("Average Assignment Score:",
      df["Assignment_Score"].mean())

print("Average Previous Score:",
      df["Previous_Score"].mean())

print("Average Final Score:",
      df["Final_Score"].mean())



print("\n========== HIGHEST FINAL SCORE ==========")

highest = df.loc[df["Final_Score"].idxmax()]
print(highest)



print("\n========== LOWEST FINAL SCORE ==========")

lowest = df.loc[df["Final_Score"].idxmin()]
print(lowest)



print("\n========== ATTENDANCE GREATER THAN 80 ==========")

print(df[df["Attendance"] > 80])

print("\n========== FINAL SCORE GREATER THAN 80 ==========")

print(df[df["Final_Score"] > 80])



print("\n========== STATISTICAL SUMMARY ==========")

print(df.describe())
```

part C: NumPy Numerical Analysis

```python
final_scores = df["Final_Score"].to_numpy()

print("Final Scores:")
print(final_scores)

print("Mean:", np.mean(final_scores))
print("Median:", np.median(final_scores))
print("Maximum:", np.max(final_scores))
print("Minimum:", np.min(final_scores))
print("Standard Deviation:", np.std(final_scores))

print("\nInterpretation:")
print("Standard deviation shows how much the final scores vary")
print("from the average final score.")
```

Part F – Calculus: Differentiation

```python
print("Given:")
print("F(x) = 40 + 5x - 0.2x^2")

print("\nDerivative:")
print("F'(x) = 5 - 0.4x")

# F'(x) = 0
x = 5 / 0.4

print("\nValue of x when F'(x) = 0:")
print(x)

print("\nInterpretation:")
print("The derivative becomes zero at 12.5 study hours.")
print("This represents the point where the model's predicted")
print("final score stops increasing and reaches its maximum.")


# Verify values
def F(x):
    return 40 + 5*x - 0.2*x**2


def dF(x):
    return 5 - 0.4*x


print("\nStudy Hours | Predicted Score | Derivative")

for x in range(1, 11):
    print(x, "          ", F(x), "          ", dF(x))
```

Part G – Calculus: Integration

```python
print("Given:")
print("L(x) = 5x - 0.2x^2")

print("\nIntegral:")
print("∫(5x - 0.2x^2) dx")
print("= 2.5x^2 - (0.2/3)x^3")

# Evaluate from 0 to 5

upper = 2.5 * (5**2) - (0.2/3) * (5**3)
lower = 0

result = upper - lower

print("\nIntegral from 0 to 5:")
print(result)


# Numerical approximation using NumPy
x_values = np.linspace(0, 5, 1000)
y_values = 5*x_values - 0.2*x_values**2

numerical_result = np.trapezoid(y_values, x_values)

print("\nNumerical approximation:")
print(numerical_result)

print("\nInterpretation:")
print("The area under the curve represents the total")
print("learning benefit accumulated from 0 to 5 study hours.")


# Plot learning benefit
plt.figure(figsize=(7, 5))
plt.plot(x_values, y_values)
plt.xlabel("Study Hours")
plt.ylabel("Learning Benefit")
plt.title("Learning Benefit Curve")
plt.grid()
plt.show()
```

Part I – Mini Prediction Model Using Linear Algebra

```python
X = df[[
    "Study_Hours",
    "Attendance",
    "Assignment_Score"
]].values

Y = df["Final_Score"].values

# Add column of 1s for intercept
X = np.column_stack((np.ones(len(X)), X))

# Calculate beta
beta = np.linalg.inv(X.T @ X) @ X.T @ Y

print("\nCoefficients:")
print(beta)

print("\nInterpretation:")
print("Beta 0 =", beta[0])
print("Beta 1 =", beta[1])
print("Beta 2 =", beta[2])
print("Beta 3 =", beta[3])

print("\nModel:")
print(
    "Final Score =",
    beta[0],
    "+",
    beta[1],
    "* Study Hours +",
    beta[2],
    "* Attendance +",
    beta[3],
    "* Assignment Score"
)

```
