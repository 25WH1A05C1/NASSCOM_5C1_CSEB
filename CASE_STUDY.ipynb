```python
# Battery sensor data
battery_data = [
    (0, 32, 400, 20, 95),
    (1, 35, 395, 22, 88),
    (2, 38, 390, 25, 80),
    (3, 40, 385, 28, 70),
    (4, 42, 380, 30, 60),
    (5, 45, 375, 32, 48)
]

# 1. Display sensor readings
print("Battery Sensor Readings:")

for data in battery_data:
    print(data)

# 2. Identify high temperature
print("\nTemperature Status:")

for data in battery_data:
    time, temperature, voltage, current, charge = data

    if temperature > 40:
        print(f"Time {time} h: High Temperature Alert")
    else:
        print(f"Time {time} h: Temperature Normal")

# 3. Identify low battery
print("\nBattery Charge Status:")

for data in battery_data:
    time, temperature, voltage, current, charge = data

    if charge < 50:
        print(f"Time {time} h: Low Battery Alert")
    else:
        print(f"Time {time} h: Battery Charge Normal")

# 4. Battery status function
def battery_status(charge):
    if charge > 70:
        return "High"
    elif 50 <= charge <= 70:
        return "Medium"
    else:
        return "Low"

print("\nBattery Status:")

for data in battery_data:
    time, temperature, voltage, current, charge = data
    print(f"Time {time} h: {battery_status(charge)}")
```

```python
import sympy as sp

t = sp.symbols('t')

C = 100 - 2*t - sp.Rational(1, 2)*t**2

dC_dt = sp.diff(C, t)

print("C(t) =", C)
print("dC/dt =", dC_dt)
```

```python
rate_t2 = dC_dt.subs(t, 2)
rate_t5 = dC_dt.subs(t, 5)

print("Rate at t=2:", rate_t2)
print("Rate at t=5:", rate_t5)
```
